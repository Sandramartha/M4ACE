```python
import pandas as pd
```

```python
excel_df = pd.read_excel("world_population_excel_workbook.xlsx")
json_df = pd.read_json("json_sample.json")
csv_df = pd.read_csv(
    "https://raw.githubusercontent.com/AlexTheAnalyst/PandasYouTubeSeries/main/countries%20of%20the%20world.csv"
)
txt_df = pd.read_csv("countries of the world.txt")

print(excel_df.head())
print(json_df.head())
print(csv_df.head())
print(txt_df.head())
```

```python
print(csv_df.isnull().sum())

numeric_columns = csv_df.select_dtypes(include="number").columns
csv_df[numeric_columns] = csv_df[numeric_columns].fillna(csv_df[numeric_columns].mean())

text_columns = csv_df.select_dtypes(include="object").columns
csv_df[text_columns] = csv_df[text_columns].fillna("Unknown")

print("Duplicates:", csv_df.duplicated().sum())
csv_df = csv_df.drop_duplicates()
```

```python
print(list(csv_df.columns))
csv_df.columns = csv_df.columns.str.strip()
print(csv_df.columns)
filtered_df = csv_df[csv_df["Population"] > 10000000]
print(filtered_df)

grouped_df = csv_df.groupby("Region")["Population"].mean()
print(grouped_df)

csv_df.to_csv("cleaned_countries_of_the_world.csv", index=False)
print("Saved cleaned dataset.")
```
