## Part 1: USSD Banking Application

```
balance=10000
PIN='1234'

def check_balance():
    print(f'Your balance is ₦{balance}')

def deposit():
    global balance
    amt=float(input('Enter deposit amount: '))
    balance+=amt
    print('Deposit successful.')

def withdraw():
    global balance
    amt=float(input('Enter withdrawal amount: '))
    if amt<=balance:
        balance-=amt
        print('Withdrawal successful.')
    else:
        print('Insufficient balance.')

def transfer():
    global balance
    acct=input('Recipient account: ')
    amt=float(input('Amount: '))
    if amt<=balance:
        balance-=amt
        print(f'₦{amt} transferred to {acct}')
    else:
        print('Insufficient balance.')

entered=input('Enter PIN: ')
if entered==PIN:
    while True:
        print('\nABC BANK *123#')
        print('1.Check Balance')
        print('2.Deposit')
        print('3.Withdraw')
        print('4.Transfer')
        print('5.Exit')
        option=input('Choose option: ')
        if option=='1':
            check_balance()
        elif option=='2':
            deposit()
        elif option=='3':
            withdraw()
        elif option=='4':
            transfer()
        elif option=='5':
            print('Thank you for banking with ABC Bank.')
            break
        else:
            print('Invalid option')
else:
    print('Incorrect PIN')
```
